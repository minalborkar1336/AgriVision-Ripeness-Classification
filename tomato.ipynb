```python
from ultralytics import YOLO
from pathlib import Path
model = YOLO('best.pt')
print(model.names)
```

## Evaluate on the cleaned test dataset
Set `DATA_YAML` to the location of your cleaned dataset's `data.yaml`. The dataset folder is not included in this results ZIP.


```python
DATA_YAML = 'tomato_clean/data.yaml'  # Change this if your dataset is stored elsewhere
metrics = model.val(data=DATA_YAML, split='test', imgsz=640, plots=True)
print('Precision:', metrics.box.mp)
print('Recall:', metrics.box.mr)
print('mAP50:', metrics.box.map50)
print('mAP50-95:', metrics.box.map)
```

## Predict on an image
Place an image in the project folder and update `IMAGE_PATH` below.


```python
IMAGE_PATH = 'test_images/tomato5.png'
results = model.predict(source=IMAGE_PATH, conf=0.25, save=True)
print('Prediction saved by Ultralytics in the runs folder.')
```
